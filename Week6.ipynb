# 2. Weekly graph question

Suppose you wanted to show the nitrate and phosphate level in a water sample.  What are the advantages and disadvantages of showing this as an area plot, as opposed to two separate line graphs?  How would you adjust the graphs shown to improve the presentation?

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series_2 =  100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series_2)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

```python
df = pd.DataFrame({'ser1': time_series, 'ser2': time_series_2}, index = datearray)
df.plot.area()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("diabetic_data.csv")

numeric_df = df.select_dtypes(include=['int64','float64'])

corr = numeric_df.corr()

plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.scatterplot(data=df,
                x='num_lab_procedures',
                y='time_in_hospital')

plt.title("Lab Procedures vs Hospital Stay")
plt.show()
```

```python
sns.pairplot(numeric_df.sample(500))
plt.show()
```

```python
life = pd.read_csv("Life Expectancy Data.csv")

life_numeric = life.select_dtypes(include=['int64','float64'])

corr = life_numeric.corr()

plt.figure(figsize=(12,10))
sns.heatmap(corr,
            cmap="coolwarm",
            center=0)

plt.title("Life Expectancy Correlation Matrix")
plt.show()
```

```python
sns.scatterplot(
    data=life,
    x="Schooling",
    y="Life expectancy "
)

plt.title("Schooling vs Life Expectancy")
plt.show()
```

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 86-98 of the Storytelling With Data book as best you can.  (The second half of chapter three).  You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt

years = [2015,2016,2017,2018,2019,2020,2021]
life = [68,69,70,71,72,73,74]

plt.figure(figsize=(8,5))

plt.plot(years,
         life,
         color='lightgray',
         linewidth=2)

plt.plot(years,
         life,
         color='blue',
         linewidth=4)

plt.title("Life Expectancy Trend")
plt.ylabel("Years")
plt.xlabel("Year")

plt.grid(alpha=0.3)

plt.show()
```
