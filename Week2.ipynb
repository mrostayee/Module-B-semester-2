**Conclusions**

The two datasets are usable for analysis, but some preprocessing is needed. I checked for duplicate rows, missing values, categorical variables, and inconsistent data. The Life Expectancy dataset contains multiple years for each country, so repeated country names are expected and should not automatically be removed.

The datasets contain both numerical and categorical variables. Numerical variables can be analyzed using the mean, median, and mode. Categorical variables may need to be encoded before they can be used in some machine learning models.

I also checked the categories to see if there is class imbalance. Before using the data for machine learning, I would clean missing or inconsistent values and encode the appropriate categorical variables.


```python
import matplotlib.pyplot as plt

sample = life.head(10)

plt.bar(sample["Country"], sample["Life expectancy "])

plt.xticks(rotation=90)
plt.ylabel("Life Expectancy")
plt.title("Life Expectancy for Selected Countries")

plt.show()
```
