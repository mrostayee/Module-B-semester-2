Storytelling With Data Graph Explanation

For this graph, I reproduced the general look and feel of a simple bar chart from the second half of Chapter 2 of Storytelling With Data. I used the Life Expectancy dataset and selected the 10 countries with the highest life expectancy.

I chose a horizontal bar chart because it makes it easy to compare the countries and quickly identify which countries have the highest values. I also added a clear title, labeled the x-axis, and reduced unnecessary visual clutter. The goal was to make the main message easy for the reader to understand without needing to study the graph for a long time.

```python
import matplotlib.pyplot as plt

top_10 = life.nlargest(10, "Life expectancy")
top_10 = top_10.sort_values("Life expectancy")

plt.figure(figsize=(10, 6))

plt.barh(
    top_10["Country"],
    top_10["Life expectancy"]
)

plt.title(
    "The countries with the highest life expectancy",
    loc="left",
    fontsize=16
)

plt.xlabel("Life expectancy (years)")

plt.grid(axis="x", alpha=0.2)

plt.tight_layout()
plt.show()
```
