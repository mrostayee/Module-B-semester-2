# 2. Weekly graph question

Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

I recreated a graph following the principles presented in Chapter 4 of Storytelling With Data. The graph emphasizes the most important data series using a strong color while the remaining information is shown in muted gray tones. Unnecessary chart elements such as excessive gridlines and distracting colors were removed to improve readability. This design helps direct the audience's attention toward the key message and makes the graph easier to interpret.

```python
import matplotlib.pyplot as plt

years = [2018, 2019, 2020, 2021, 2022, 2023]
values = [60, 63, 68, 74, 80, 88]

plt.figure(figsize=(8,5))

plt.plot(years, values, color='steelblue', linewidth=3)
plt.scatter(years[-1], values[-1], color='red', s=100)

plt.title("Example Storytelling With Data Chart")
plt.xlabel("Year")
plt.ylabel("Value")

plt.grid(alpha=0.2)
plt.show()
```
